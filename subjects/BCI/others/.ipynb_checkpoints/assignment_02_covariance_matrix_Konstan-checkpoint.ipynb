## Exercise 2: covariance estimator for n-dimensional data
The function `generate_covariance()` that generates a covariance matrix. The resulting covariance matrix is used with `generate_data()` to generate data.
1. Implement a function `estimate_covariance(X)` that calculates the covariance matrix for a given data set $\mathbf{X}\in\mathbb{R}^{N \times D}$ of $N$ examples by $D$ features.
1. Test, compare and validate your function with `np.cov` (read its documentation) and the true covariance matrix `cov_true`.
1. Visualize the three covariance matrices (e.g., `plt.imshow()`) using separate subplots (`plt.subplots()`). 

**Hint**: inspect the shape of your output using `var.shape` and compare it with your expectations for a desired input of $N \times D$.

```python
def generate_covariance(D):
    tmp = np.random.randn(D, D)
    cov = np.dot(tmp, tmp.transpose()) 
    return cov
```

```python
np.random.seed(42) # ensures replicable results
N = 200
D = 100
cov_true = generate_covariance(D)
X = generate_data(N, cov_true, np.zeros(D))
```

```python
# Answer here
def estimate_covariance(X):
    mean = np.mean(X, axis=0) #mean for every feature(column). here it will be 100 mean values for the 100 different features
    X_centered = X - mean 
    X_centered_transpose = X_centered.transpose()
    product = X_centered_transpose @ X_centered
    covariance = product / (X.shape[0] - 1)
    return covariance

print("np.cov")
print(np.cov(X, rowvar=False, bias=False)) #Default normalization (False) is by (N - 1), where N is the number of observations given (unbiased estimate).
print("\n")
print("estimate_covariance")
print(estimate_covariance(X))
print("\n")
print("cov_true")
print(cov_true)

#plt.imshow(cov_true)
#plt.figure() #new plot
#plt.imshow(estimate_covariance(X))
#plt.figure()
#plt.imshow(np.cov(X, rowvar=False, bias=False))

print("Shape of my covariance matrix: ",estimate_covariance(X).shape) #must be DxD

fig, axes = plt.subplots(1, 3)
axes[0].set_title("cov_true")
axes[1].set_title("estimate_covariance")
axes[2].set_title("np.cov")
axes[0].imshow(cov_true)
axes[1].imshow(estimate_covariance(X))
axes[2].imshow(np.cov(X, rowvar=False, bias=False))
    
```

## Exercise 5:  Improving covariance estimation with shrinkage regularization
You have observed a systematic error in the eigenvalue estimates, even for $N > D$. We can compensate by using shrinkage regularization.

To work with shrinkage regularization we can use the `shrunk_covariance()` function from the "sklearn" library. See `estimate_regularized_covariance()` that uses that regularization.

1. Estimate the empirical covariance matrix with and without regularization;
1. For the *true*, *empirical*, and *regularized* covariance matrices, compute the eigenvalue spectrum;
1. Plot the eigenvalue spectra in a single figure;
1. Compute the mean squared error (MSE) between the true and estimated covariance matrices;
1. Report your observations.

```python
def estimate_regularized_covariance(X, shrink_param=0.3):
    cov = np.cov(X.T)
    regularized_cov = shrunk_covariance(cov, shrink_param)
    return regularized_cov
```

```python
np.random.seed(42) # ensures replicable results
N = 120
cov_true = generate_covariance(D)
X = generate_data(N, cov_true, np.zeros(D))
```

```python
# Answer here
estimated_covariance = estimate_covariance(X)
estimated_regularized_covariance = estimate_regularized_covariance(X)

evalue_cov_true, evect_cov_true = np.linalg.eig(cov_true)
evalue_estimated_covariance, evect_estimated_covariance = np.linalg.eig(estimated_covariance)
evalue_estimated_regularized_covariance, evect_estimated_regularized_covariance = np.linalg.eig(estimated_regularized_covariance)

evalue_cov_true = -np.sort(-evalue_cov_true)
evalue_estimated_covariance = -np.sort(-evalue_estimated_covariance)
evalue_estimated_regularized_covariance = -np.sort(-evalue_estimated_regularized_covariance)

plt.plot(evalue_cov_true, label="cov_true")
plt.plot(evalue_estimated_covariance, label="estimated_covariance")
plt.plot(evalue_estimated_regularized_covariance, label="estimated_regularized_covariance")
plt.xlabel("eigenvalues descending ranking")
plt.ylabel("eigenvalue")
plt.legend() #shows the labels

mse_with_estimated_covariance = np.mean((cov_true - estimated_covariance)**2)
mse_with_estimated_regularized_covariance = np.mean((cov_true - estimated_regularized_covariance)**2) 
print("MSE between true covariance and estimated covariance: ", mse_with_estimated_covariance)
print("MSE between true covariance and estimated regularized covariance: ", mse_with_estimated_regularized_covariance)

'''
5. As expected, MSE in the case of the estimated regularized covariance is lower than the MS in the case of the estimated covariance. 
That shows that shrinkage regularization can provide a better estimate of the true covariance matrix.
We can also observe that regularization reduces the largest eigenvalues and increases the smallest eigenvalues, making them less extreme.
'''
```
